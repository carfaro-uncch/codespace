```r
library('tidyverse')
```

```r
llm = read.csv('https://raw.githubusercontent.com/UNC-BIOS-512/jupyterlite/refs/heads/main/data/llm_usage.csv')
llm |> head()
```

```r
llm |>
  mutate(date = as.Date(date)) |>
  group_by(date) |>
  summarise(cost_usd = sum(cost_usd)) |>
  mutate(cost_usd = cumsum(cost_usd)) |>
  print(n = Inf)
```

```r
llm |>
  mutate(date = as.Date(date)) |>
  group_by(date) |>
  summarise(cost_usd = sum(cost_usd)) |>
  mutate(cost_usd = cumsum(cost_usd)) |>
  ggplot(aes(x = date, y = cost_usd)) +
  geom_line() +
  geom_point() +
  labs(
    x = "Date",
    y = "Cumulative cost (USD)",
    title = "Cost over time"
  )
```
