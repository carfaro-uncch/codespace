```r
library('tidymodels')
library('readr')
```

```r
gss = read_csv(https://github.com/UNC-BIOS-512/jupyterlite/blob/main/data/gss_subset.csv)
gss |> glimpse()
```

```r
mod = linear_reg()

mod
```

```r
mod_fit = mod |> fit(realinc ~ educ, data = gss)

mod_fit
```

```r
options(repr.plot.res = 250, repr.plot.height = 4, repr.plot.width = 8)
```

```r
# plot some predictions
gss2 = predict(mod_fit, gss) |>
    bind_cols(gss) 

gss2 |> glimpse()
```

```r
ggplot(gss2, aes(x = educ, y = realinc)) + 
    geom_point(alpha = 0.1, position = position_jitter(width = 0.1)) +
    geom_line(aes(x = educ, y = .pred), linewidth = 2, color = 'coral')
```
